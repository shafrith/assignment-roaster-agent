```python
def chat(q):
    r = openai.chat.completions.create(model="gpt-4o", messages=[{"role": "user", "content": "You are a helpful assistant. Answer this: " + q}])
    print(r)
    return r.choices[0].message.content
```

```python
try:
    ans = chat('what is the best product?')
except:
    pass
print(ans)
```

```python
for q in ['hi', 'hello', 'how are you']:
    print(chat(q))
```
